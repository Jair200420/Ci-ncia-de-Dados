# Condicionais em Python

Em um programa, temos instruções sequenciais. Por vezes, desejamos que algumas instruções sejam executadas apenas quando alguma condição é satisfeita. Nestes casos, falamos de ferramentas de controle de fluxo condicionais que, no Python, utilizam as palavras reservadas `if`, `elif` e `else`. Vejamos alguns exemplos:

## If

Caso mais simples, com a sintaxe:

```python
if <expressão lógica>:
    <instrução>
```

Onde `expressão lógica` representa nossa condição, algo que possa ser verdadeiro ou falso.

````python
# Exemplo: verificar se um número é positivo
numero = -5

if numero >= 0:
    print(f"{numero} é positivo")
else:
    print(f"{numero} não é positivo")
````

````python
# Exemplo: bonificação de funcionários

funcionario = {
    'nome': 'João',
    'idade': 25,
    'salario': 6000.0,
    'cargo': 'Cientista de Dados',
    'tempo_de_casa': 8,
}

if funcionario["tempo_de_casa"] >= 5:
    funcionario["salario"] *= 1.2
    print("O funcionário recebeu uma bonificação de 20%")
````

## if / else

````python
novo_numero = 8

if novo_numero % 2 ==0: # "%" simboliza o resto da divisão
    print(f"{novo_numero} é par")
else:
    print(f"{novo_numero} é ímpar")
````

## if / elif/ else

````python
nota = 83

if nota >= 90:
    print("O aluno recebeu conceito A")
elif nota >= 80:
    print("O aluno recebeu conceito B")
elif nota >= 70:
    print("O aluno recebeu conceito C")
elif nota>= 60:
    print("O aluno recebeu conceito D")
else:
    print("O aluno recebeu conceito F")
````

````python
if (nota < 0) or (nota > 100):
    print("Nota inválida.")
else:
    if nota >= 90:
        print("O aluno recebeu conceito A")
    elif nota >= 80:
        print("O aluno recebeu conceito B")
    elif nota >= 70:
        print("O aluno recebeu conceito C")
    elif nota>= 60:
        print("O aluno recebeu conceito D")
    else:
        print("O aluno recebeu conceito F")
````
