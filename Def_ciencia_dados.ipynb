**Definição de ciência de dados:**

- Ciência da computação;
- Métodos científicos;
- Estatística;
- Algoritmos;
- Processos e Sistemas.

Utilizar essas áreas para extrair ou extrapolar conhecimento a partir de dados diversos. Sendo eles, potencialmente, ruidosos, estruturados ou não-estruturados.

A ciência de dados extrai informações e extrapola os conhecimentos, a partir da tecnologia e modelos matemáticos, para solucionar problemas reais.

- Softwares;
- Machine learning;
- Estatística.

Método científico:

- Obrservação;
- Hipóteses;
- Experimento;
- Análise / Resultado;
- Conclusão.

**Ciência de dados baseado no método científico**

- Entendimento do negócio;
- Entendimento dos dados;
- Preparação dos dados;
- Análise e Modelagem;
- Validação;
- Produção.

**Dados x Inteligência**

- Dados: é o registro de algo que aconteceu;
- Informação: é a contextualização do dado. O dado passou por um processamento;
- Conhecimento: fornecer um significado para o dado. Análise + entendimento;
- Inteligência (de negócios): utilizar o conhecimento para tomar uma decisão para melhora.

**Extrair x Extrapolar conhecimento**

- Extrair: Processar e analisar os dados para encontrar informações diretas que ja estão presentes, mas não diretamente óbvias;
- Extrapolar: Utilizar padrões encontrados nos dados para afzer previsões sobre dados que ainda não vimos.
