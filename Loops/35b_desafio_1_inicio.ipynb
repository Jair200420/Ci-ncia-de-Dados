# Desafio

Escreva um programa em Python que verifique se uma senha fornecida como uma string cumpre os seguintes critérios de segurança:

- Tem pelo menos 8 caracteres.
- Contém pelo menos uma letra maiúscula.
- Contém pelo menos uma letra minúscula.
- Contém pelo menos um número.
- Contém pelo menos um caractere especial (considere caracteres especiais como `!@#$%^&*()-_=+`).

Se a senha atender a todos os critérios, o programa deve imprimir `"Senha válida"`. Caso contrário, deve imprimir `"Senha inválida"`.

Dica: pesquise pelos métodos  `isupper`, `islower`, `isdigit` de strings.

```python
# Definindo os caracteres especiais
caracteres_especiais = "!@#$%^&*()-_=+"

nova_senha = input("Digite uma nova senha: ")

# Inicializando as variáveis de verificação
tem_maiuscula = False
tem_minuscula = False
tem_numero = False
tem_especial = False

# Verificando as condições
if len(nova_senha) >= 8:  # Primeiro checa se a senha tem pelo menos 8 caracteres
    for caracter in nova_senha:
        if caracter.isupper():
            tem_maiuscula = True
        elif caracter.islower():
            tem_minuscula = True
        elif caracter.isdigit():
            tem_numero = True
        elif caracter in caracteres_especiais:
            tem_especial = True

    # Verificando se todas as condições foram atendidas
    if tem_maiuscula and tem_minuscula and tem_numero and tem_especial:
        print("Senha válida... Sua senha foi registrada.")
    else:
        print("Senha inválida... Verifique se sua senha contém pelo menos uma letra maiúscula, uma letra minúscula, um número e um caractere especial (!@#$%^&*()-_=+).")
else:
    print("Senha inválida... A senha deve conter pelo menos 8 caracteres.")
```
