# Pandas básico

```python
import pandas as pd

BASE = "dados/cereal_sem_negativos.csv"

df = pd.read_csv(BASE)

df.head()
```

```python
df.info()
```

```python
df["mfr"].value_counts().plot(kind="bar")
```

```python
df["mfr"].value_counts().plot(
    kind="bar",
    xlabel= "Fabricante",
    ylabel= "Quantidade",
    title= "Contagem de cereais por fabricante"
    )
```

```python
df.groupby("mfr")["calories"].mean().plot(
    kind="bar",
    xlabel="Fabricante",
    ylabel="Calorias",
    title="Calorias por fabricante"
)
```
