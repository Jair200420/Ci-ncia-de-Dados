```python
import speech_recognition as sr
import time
```

```python
# Reconhecedor:

rec = sr.Recognizer()
```

```python
acabou = False

def tratar_audio(rec, audio):
    global acabou # É a mesma variável em todo o código. Variáveis dentro e fora de uma função são diferentes
    
    try:
        texto = rec.recognize_amazon(audio, language="pt-BR")
        print(texto)
        if "encerrar gravação" in texto:
            acabou == True
    except:
        print("Não foi possível entender")

with sr.Microphone() as microfone:
    rec.adjust_for_ambient_noise(microfone)
    rec.pause_threshold = 0.6
    rec.dynamic_energy_adjustment_ratio = 2.0 # Testar valor para melhor reconhecimento
    print("Pode começar a falar")
    audio = rec.listen(microfone)

parar_ouvir = rec.listen_in_background(microfone, tratar_audio)

while True:
    time.sleep(0.1)
    if acabou == True:
        break
```
