### Programa de reconhecimento das mãos

```python
import cv2
import mediapipe
```

```python
# Vincular webcam no pyhthon:

webcam = cv2.VideoCapture() # Inicializar a conexão entre a webcam e o PC

# Inicializando o reconhecimento de mãos:
solucao_reconhecimento_maos = mediapipe.solutions.hands
desenho_mp = mediapipe.solutions.drawing_utils # Desenho das mãos no reconhecimento
maos = solucao_reconhecimento_maos.Hands()

if webcam.isOpened():
    validacao, frame = webcam.read() # Armazenando duas informações. Para mais detalhes, rever o vídeo
    while validacao == True:
        validacao, frame = webcam.read() # Pegar todos os frames da tela

        frameRGB = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB) # Formato padrãod a maioria dos códigos de leitura de imagem RGB (Red, Green, Blue)
        
        lista_maos = maos.process(frameRGB) # Reconhecimento das mãos
        if lista_maos.multi_hand_landmarks: # Reconhecer duas mãos
            for mao in lista_maos.multi_hand_landmarks:
                desenho_mp.draw_landmarks(frame, mao, solucao_reconhecimento_maos.HAND_CONNECTIONS)

        cv2.imshow("Vídeo Webcam", frame) # Mostrar os frames na tela
        tecla = cv2.waitKey(2) # Controle de fps. O "2" = 30 fps. A variável, também, auxilia na parada do loop]
        if tecla == 27: # para o "OpenCV" (cv2) a tecla ESC = 27
            break        

webcam.release() # Finalizar a conexão entra a webcam e o PC
cv2.destroyAllWindows() # Fechar a janela do Webcam
```
