```python
import speech_recognition as sr
```

```python
# Reconhecedor:

rec = sr.Recognizer()
```

```python
# Áudio: Microfone:

with sr.Microphone() as microfone: # A estrutura "with" abre e fecha o microfone
    rec.adjust_for_ambient_noise(microfone) # Ajuste de ruído
    print("Pode começar a falar")
    audio = rec.listen(microfone) # Guarda o áudio captado
    try:
        texto = rec.recognize_amazon(audio, language="pt-BR") # Reconhece traduz o áudio
        print(texto)
    except:
        print("Não foi possível capturar o áudio")

with open("audio.wav", "wb") as arquivo: # Criando o arquivo de áudio
    arquivo.write(audio.get_wav_data())
```
