# Variáveis

```python
salario_base = 1500
comissao = 0.05

salario_vendas = salario_base * comissao
salario_final = salario_base + salario_vendas

print(salario_final)
```

```python
salario_base = salario_base + 200
print(salario_base)
```

```python
salario_base += 200
print(salario_base)
```

```python
salario_base *= 5

print(salario_base)
```

**Comparações**

```python
salario_base = 1500
salario_minimo = 1412
```

```python
salario_base > salario_minimo
```

```python
salario_minimo > salario_base
```

```python
salario_minimo == salario_base
```

```python
salario_base >= salario_minimo
```

```python
salario_minimo >= salario_base
```

```python
salario_minimo < salario_base < 3000
```

**Operador "AND" e "OR"**

```python
(salario_minimo < 1450) and (1450 < salario_base)
```
